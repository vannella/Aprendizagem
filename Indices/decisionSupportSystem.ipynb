# **SISTEMAS DE APOIO Á DECISÃO**
## **ETAPAS DO DESAFIO:**
### 1️⃣ **Descrição do Sistema de Apoio à Decisão**

O sistema de apoio à decisão tem como objetivo avaliar a concessão de crédito a um cliente com base em critérios previamente definidos.
A partir das informações fornecidas (**renda, pontuação de crédito, idade e histórico de pagamento**), o sistema analisa o cenário e retorna uma recomendação automática:

- **Aprovado**

- **Em revisão**

- **Negado** (Rejeitado)

Esse tipo de sistema utiliza regras de produção (**SE → ENTÃO**), muito comuns em sistemas especialistas e de apoio à decisão.

## 2️⃣ Variáveis de Entrada:


Variável do cliente baseado no mesmo exemplo do slide 04

**Renda mensal:** R$ 7.000

**Pontuação de crédito:** 800

**Idade:** 30 anos

**Histórico de pagamento:** bom

## 3️⃣ Criação da Lógica de Decisão:

A lógica segue esta ordem de prioridade:
- Idade acima de 60 anos → **Negado**

- Alta renda + bom score + bom histórico → **Aprovado**

- Combinações intermediárias → **Em revisão**

- Baixa renda + score baixo + histórico ruim → **Negado**



Para vizualização:

🌳 Árvore de Decisão – Concessão de Crédito

                         [Idade > 60?]
                         /           \

                        [Score > 700?]        [Score > 700?]
                           /     \               /     \

                     /    \                                /        \
                  Sim      Não                          Sim          Não
               Aprovado  Revisão                      Negado        Revisão


## 4️⃣ Implementação em Python

```python
# Entradas
renda = 7000
pontuacao_credito = 800
idade = 30
historico_pagamento = "bom"  # "bom" ou "ruim"

# Lógica de decisão
if idade > 60:
    resultado = "Negado"

elif renda > 5000 and pontuacao_credito > 700 and historico_pagamento == "bom" and idade < 60:
    resultado = "Aprovado"

elif renda < 5000 and pontuacao_credito < 700 and historico_pagamento == "ruim" and idade < 60:
    resultado = "Negado"

else:
    resultado = "Em revisão"

# Saída
print("Resultado da análise de crédito:", resultado)
```
